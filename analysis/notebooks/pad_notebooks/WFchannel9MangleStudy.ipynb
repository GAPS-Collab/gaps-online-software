```python
print("mangled:    ", {n: len(rms_mangled_lists[n])    for n in bin_counts})
print("non-mangled:", {n: len(rms_nonmangled_lists[n]) for n in bin_counts})
#here
# =========================================================
# histograms of ch9 first-N-bin RMS: mangled RBs (red) vs non-mangled RBs (blue)
# =========================================================
bins = 200


ncols = 3
# +1 reserves a spare cell for the results table
nrows = int(np.ceil((len(bin_counts) + 1) / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4 * nrows))
axes = np.atleast_1d(axes).ravel()

def overlap_coefficient(a, b, edges):
    """Overlap area of two normalized histograms on shared bin edges.
    0 = no overlap (perfect split), 1 = identical distributions."""
    pa, _ = np.histogram(a, bins=edges, density=True)
    pb, _ = np.histogram(b, bins=edges, density=True)
    widths = np.diff(edges)
    # convert densities to per-bin probabilities, then sum the minima
    return float(np.sum(np.minimum(pa, pb) * widths))

overlaps = {}

for ax, n in zip(axes, bin_counts):
    mangled    = np.asarray(rms_mangled_lists[n],    dtype=float)
    nonmangled = np.asarray(rms_nonmangled_lists[n], dtype=float)
    # drop NaN/inf (e.g. IncompleteReadout RBs with too-short ch9) so they
    # don't poison the bin edges / overlap integral
    mangled    = mangled[np.isfinite(mangled)]
    nonmangled = nonmangled[np.isfinite(nonmangled)]
    # shared bin edges -> equal bin widths for both populations
    combined = np.concatenate([mangled, nonmangled])
    edges = np.linspace(combined.min(), combined.max(), bins + 1)

    ovl = overlap_coefficient(mangled, nonmangled, edges)
    overlaps[n] = ovl

    # density=True so the very different population sizes are comparable in shape
    ax.hist(nonmangled, bins=edges, color="blue", alpha=0.6, density=True, label=f"non-mangled ({nonmangled.size})")
    ax.hist(mangled,    bins=edges, color="red",  alpha=0.6, density=True, label=f"mangled ({mangled.size})")
    ax.set_xlabel("ch9 RMS (ADC)")
    ax.set_ylabel("density")
    ax.set_title(f"first {n} bins  (overlap = {ovl:.3f})")
    ax.set_yscale("log")
    ax.legend()

# window with the least overlap = best mangled vs non-mangled separation
best_n = min(overlaps, key=overlaps.get)
print("overlap coefficient per window (lower = better mangled/non-mangled split):")
for n in bin_counts:
    flag = "  <-- best" if n == best_n else ""
    print(f"  first {n:>3} bins : overlap = {overlaps[n]:.4f}{flag}")
print(f"\nBest separating window: first {best_n} bins (overlap = {overlaps[best_n]:.4f})")

# hide unused subplots, but reuse the bottom-right one for the results table
n_used = len(bin_counts)
for ax in axes[n_used:]:
    ax.set_visible(False)

if n_used < len(axes):
    results_ax = axes[-1]
    results_ax.set_visible(True)
    results_ax.axis("off")
    table_lines = ["overlap (mangled vs non-mangled)", "(lower = better split)", ""]
    for n in bin_counts:
        flag = "  <-- best" if n == best_n else ""
        table_lines.append(f"first {n:>3} bins : {overlaps[n]:.4f}{flag}")
    results_ax.text(
        0.02, 0.98, "\n".join(table_lines),
        va="top", ha="left", family="monospace", fontsize=11,
        transform=results_ax.transAxes,
    )

fig.suptitle(
    f"Channel 9 first-N-bin RMS — mangled (red) vs non-mangled (blue) RBs"
    f"\nbest split: first {best_n} bins (overlap = {overlaps[best_n]:.3f})",
    y=1.0,
)
# leave room at the top for the (2-line) suptitle so it doesn't overlap the subplots
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
